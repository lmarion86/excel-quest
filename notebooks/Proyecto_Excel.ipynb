```python
# Análisis simulado de datos de usuarios (Ejemplo de backend de pruebas)
import pandas as pd
import matplotlib.pyplot as plt

# Datos simulados de progreso (XP obtenida en pre-test vs post-test)
datos = {
    'Estudiante': ['E1', 'E2', 'E3', 'E4', 'E5'],
    'Puntaje_Pre_Test': [20, 35, 10, 50, 25],
    'Puntaje_Post_Test': [85, 90, 75, 95, 80]
}

df = pd.DataFrame(datos)
df
```

```python
# Visualizar la mejora en el aprendizaje
df.set_index('Estudiante').plot(kind='bar', figsize=(10, 6), color=['#ff5252', '#00e676'])
plt.title('Mejora de Puntaje: Pre-Test vs Post-Test')
plt.ylabel('Puntaje (%)')
plt.xlabel('Estudiantes')
plt.show()
```
